```python
import pandas as pd
import os

# Klasör yolları
base_path = "../data/03_fingerprints"
genes = ["MYC", "CTNNB1"]

# Hocanın formatına en yakın ve senin verilerine uygun sütunlar
required_columns = [
    "PUBCHEM_SID", 
    "PUBCHEM_CID", 
    "LABEL", 
    "Morgan_Fingerprint_2048", # 'Fingerprint' yerine daha spesifik bir isim
    "Standard Value",
    "Standard Type"
]

for gene in genes:
    input_file = f"{base_path}/{gene}_active.csv"
    output_file = f"{base_path}/{gene}_active_curated.csv"
    
    if os.path.exists(input_file):
        print(f"{gene} Aktif listesi işleniyor...")
        df = pd.read_csv(input_file)

        if "Fingerprint" in df.columns:
            df = df.rename(columns={"Fingerprint": "Morgan_Fingerprint_2048"})
        
    
        # 2. Sadece gerekli sütunları seç (Mevcut olanları filtrele)
        existing_cols = [col for col in required_columns if col in df.columns]
        df_filtered = df[existing_cols].copy()
        
        # 3. Aynı molekülün tekrarlarını (Duplicate) temizle
        # match kontrolü için CID'lerin benzersiz olması şart
        df_filtered = df_filtered.drop_duplicates(subset=['PUBCHEM_CID'])
        
        # 4. Kaydet
        df_filtered.to_csv(output_file, index=False)
        
        print(f"{gene} için {len(df_filtered)} benzersiz aktif hazırlandı.")
    else:
        print(f"{input_file} bulunamadı!")

print("\n Pozitif kontrol listeleri (active_curated.csv) hazır!")
```
