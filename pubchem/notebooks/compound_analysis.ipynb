```python
from rdkit import Chem
from rdkit.Chem import Descriptors, rdMolDescriptors

```

```python
smiles = "CC1=C(C(=CC=C1)C)NC(=O)C2=C3CCC4=CN=C(N=C4N3C=C2)NC5=C(C=C(C=C5)C(=O)NC6CCC(CC6)N)OC"
mol = Chem.MolFromSmiles(smiles)

print(f"MW: {Descriptors.MolWt(mol):.2f}")
print(f"TPSA: {Descriptors.TPSA(mol):.2f}")
print(f"LogP: {Descriptors.MolLogP(mol):.2f}")
print(f"Fsp3: {rdMolDescriptors.CalcFractionCSP3(mol):.2f}")
```
